```python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
)

# Work with a copy of the corrected dataset
data = m.copy()

# Create features available at submission time
data["month_cat"] = data["created_date"].dt.month.astype(str)
data["weekday_cat"] = data["created_date"].dt.dayofweek.astype(str)
data["created_hour"] = data["created_date"].dt.hour

categorical_features = [
    "problem_formerly_complaint_type",
    "problem_detail_formerly_descriptor",
    "borough",
    "incident_zip",
    "open_data_channel_type",
    "month_cat",
    "weekday_cat",
]

numeric_features = ["created_hour"]
features = categorical_features + numeric_features

# Prevent missing categorical values from causing encoding issues.
for col in categorical_features:
    data[col] = (
        data[col]
        .astype("object")
        .where(data[col].notna(), "Unknown")
    )

# Chronological split
year = data["created_date"].dt.year

train = data.loc[year.between(2020, 2023)].copy()
validation = data.loc[year.eq(2024)].copy()
test = data.loc[year.eq(2025)].copy()
stress_test = data.loc[year.eq(2026)].copy()

print("Training:", len(train))
print("Validation:", len(validation))
print("Historical test:", len(test))
print("2026 stress test:", len(stress_test))

# Preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

preprocessor = ColumnTransformer([
    ("categorical", categorical_pipeline, categorical_features),
    ("numeric", numeric_pipeline, numeric_features),
])

# Baseline model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight=None
    )),
])

X_train = train[features]
y_train = train["delay_gt_7d"].astype(int)

model.fit(X_train, y_train)

# Evaluate each period without retraining
def evaluate(name, split):
    if split.empty:
        print(f"\n{name}: no data")
        return

    X = split[features]
    y = split["delay_gt_7d"].astype(int)
    probs = model.predict_proba(X)[:, 1]

    print(f"\n--- {name} ---")
    print("Rows:", f"{len(split):,}")
    print("Actual delay rate:", round(y.mean(), 4))
    print(
        "PR-AUC:",
        round(average_precision_score(y, probs), 4)
    )
    print(
        "ROC-AUC:",
        round(roc_auc_score(y, probs), 4)
    )
    print(
        "Brier score:",
        round(brier_score_loss(y, probs), 4)
    )
    print(
        "Constant-baseline Brier:",
        round(
            brier_score_loss(
                y,
                np.full(len(y), y_train.mean())
            ),
            4
        )
    )

evaluate("Validation — 2024", validation)
evaluate("Final historical test — 2025", test)
evaluate("Distribution-shift stress test — 2026", stress_test)
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import brier_score_loss

features = categorical_features + enhanced_numeric


def inspect_predictions(name, split, model):
    result = split[
        ["created_date", "delay_gt_7d"]
    ].copy()

    result["predicted_risk"] = model.predict_proba(
        split[features]
    )[:, 1]

    print(f"\n--- {name} ---")
    print("Actual delay rate:",
          round(result["delay_gt_7d"].mean(), 4))
    print("Average predicted risk:",
          round(result["predicted_risk"].mean(), 4))
    print("Brier score:",
          round(brier_score_loss(
              result["delay_gt_7d"],
              result["predicted_risk"]
          ), 4))

    # Group requests into ten risk bands.
    result["risk_band"] = pd.qcut(
        result["predicted_risk"],
        q=10,
        duplicates="drop"
    )

    calibration = result.groupby(
        "risk_band", observed=True
    ).agg(
        requests=("delay_gt_7d", "size"),
        average_predicted_risk=("predicted_risk", "mean"),
        actual_delay_rate=("delay_gt_7d", "mean")
    )

    print("\nPredicted versus observed risk:")
    print(calibration.round(3).to_string())

    return result


pred_2025 = inspect_predictions(
    "2025 historical test",
    test,
    zip_city_model
)

pred_2026 = inspect_predictions(
    "2026 stress test",
    stress_test,
    zip_city_model
)

# Check which months drive the 2026 result.
pred_2026["month"] = (
    pred_2026["created_date"].dt.to_period("M")
)

print("\n2026 MONTHLY CALIBRATION")
print(
    pred_2026.groupby("month").agg(
        requests=("delay_gt_7d", "size"),
        actual_delay_rate=("delay_gt_7d", "mean"),
        average_predicted_risk=("predicted_risk", "mean")
    ).round(3).to_string()
)
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import brier_score_loss, roc_auc_score

score_features = categorical_features + enhanced_numeric

def score_period(split):
    result = split.copy()

    result["predicted_risk"] = zip_city_model.predict_proba(
        result[score_features]
    )[:, 1]

    result["status_clean"] = (
        result["status"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    result["is_duplicate_resolution"] = (
        result["resolution_description"]
        .fillna("")
        .str.contains(
            r"duplicate|previously filed complaint",
            case=False,
            regex=True,
        )
    )

    result["month"] = result["created_date"].dt.to_period("M")

    return result


scored_2025 = score_period(test)
scored_2026 = score_period(stress_test)

# 1. Monthly prediction quality and status composition
all_scored = pd.concat(
    [scored_2025, scored_2026],
    ignore_index=True,
)

monthly = all_scored.groupby("month").agg(
    requests=("unique_key", "size"),
    actual_delay_rate=("delay_gt_7d", "mean"),
    predicted_risk=("predicted_risk", "mean"),
    duplicate_share=("is_duplicate_resolution", "mean"),
    closed_share=(
        "status_clean",
        lambda s: s.eq("CLOSED").mean()
    ),
)

monthly["brier_score"] = all_scored.groupby("month").apply(
    lambda g: brier_score_loss(
        g["delay_gt_7d"],
        g["predicted_risk"]
    ),
    include_groups=False,
)

print("MONTHLY DIAGNOSTICS")
print(monthly.loc["2025-01":].round(3).to_string())

# 2. Compare duplicate and non-duplicate cases after closure
closed_scored = all_scored.loc[
    all_scored["status_clean"].eq("CLOSED")
    & all_scored["closed_date"].notna()
].copy()

print("\nCLOSED REQUESTS: DUPLICATE VS NON-DUPLICATE")

print(
    closed_scored.groupby("is_duplicate_resolution").agg(
        requests=("unique_key", "size"),
        delay_rate=("delay_gt_7d", "mean"),
        average_predicted_risk=("predicted_risk", "mean"),
        median_closure_days=("closure_days", "median"),
    ).round(3).to_string()
)

# 3. Compare feature distributions across periods
workload_features = [
    "borough_reports_prev_7d",
    "borough_reports_prev_28d",
    "zip_reports_prev_7d",
    "zip_reports_prev_28d",
    "city_reports_prev_7d",
    "city_reports_prev_28d",
]

print("\nWORKLOAD FEATURE DISTRIBUTIONS")
for name, split in [
    ("Train", enhanced[
        enhanced["created_date"].dt.year.between(2020, 2023)
    ]),
    ("2025 test", scored_2025),
    ("2026 stress test", scored_2026),
]:
    print(f"\n{name}")
    print(
        split[workload_features]
        .quantile([0.50, 0.90, 0.99])
        .round(1)
        .to_string()
    )
```

```python
import time
import warnings
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
    log_loss,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

warnings.filterwarnings("ignore", category=UserWarning)

# =====================================================
# 1. PREPARE THE COMMON DATASET
# =====================================================

bench = enhanced.copy()

bench["created_date"] = pd.to_datetime(
    bench["created_date"], errors="coerce"
)

bench["delay_gt_7d"] = pd.to_numeric(
    bench["delay_gt_7d"], errors="coerce"
)

bench = bench.dropna(
    subset=["created_date", "delay_gt_7d"]
).copy()

bench["created_month_cat"] = (
    bench["created_date"].dt.month.astype(str)
)
bench["created_weekday_cat"] = (
    bench["created_date"].dt.dayofweek.astype(str)
)
bench["created_hour"] = bench["created_date"].dt.hour

# Ensure categoricals exist
if "borough_clean" not in bench.columns:
    bench["borough_clean"] = (
        bench["borough"].astype("string").fillna("Unknown")
    )

categorical_candidates = [
    "problem_formerly_complaint_type",
    "problem_detail_formerly_descriptor",
    "borough_clean",
    "incident_zip",
    "open_data_channel_type",
    "created_month_cat",
    "created_weekday_cat",
]

categorical_features = [
    c for c in categorical_candidates if c in bench.columns
]

numeric_candidates = [
    "created_hour",
    "borough_reports_prev_7d",
    "borough_reports_prev_28d",
    "zip_reports_prev_7d",
    "zip_reports_prev_28d",
    "city_reports_prev_7d",
    "city_reports_prev_28d",
]

numeric_features = [
    c for c in numeric_candidates if c in bench.columns
]

for col in categorical_features:
    bench[col] = (
        bench[col].astype("string").fillna("Unknown").astype(str)
    )

for col in numeric_features:
    bench[col] = pd.to_numeric(
        bench[col], errors="coerce"
    )

features = categorical_features + numeric_features
target = "delay_gt_7d"

print("Features:", features)
print("Total labeled rows:", f"{len(bench):,}")

# =====================================================
# 2. CHRONOLOGICAL SPLITS
# =====================================================

year = bench["created_date"].dt.year

train = bench.loc[year.between(2020, 2023)].copy()
validation = bench.loc[year.eq(2024)].copy()
test = bench.loc[year.eq(2025)].copy()
stress_test = bench.loc[year.eq(2026)].copy()

X_train, y_train = train[features], train[target].astype(int)

splits = {
    "Validation 2024": validation,
    "Test 2025": test,
    "Stress 2026": stress_test,
}

print("\nSplit sizes:")
print({
    "train": len(train),
    "validation": len(validation),
    "test": len(test),
    "stress_test": len(stress_test),
})

# =====================================================
# 3. SHARED PREPROCESSING
# =====================================================

def make_preprocessor():
    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ])

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ])

    return ColumnTransformer([
        ("cat", categorical_pipeline, categorical_features),
        ("num", numeric_pipeline, numeric_features),
    ])

def make_pipeline(classifier):
    return Pipeline([
        ("preprocessor", make_preprocessor()),
        ("classifier", classifier),
    ])

# =====================================================
# 4. DEFINE ALL CANDIDATE MODELS
# =====================================================

models = {
    "Dummy baseline": DummyClassifier(strategy="prior"),

    "Logistic Regression": make_pipeline(
        LogisticRegression(max_iter=1000, C=1.0)
    ),

    "Random Forest": make_pipeline(
        RandomForestClassifier(
            n_estimators=100,
            max_depth=18,
            min_samples_leaf=30,
            max_features="sqrt",
            n_jobs=-1,
            random_state=42,
        )
    ),

    "Extra Trees": make_pipeline(
        ExtraTreesClassifier(
            n_estimators=100,
            max_depth=20,
            min_samples_leaf=30,
            max_features="sqrt",
            n_jobs=-1,
            random_state=42,
        )
    ),
}

# Add XGBoost when installed.
try:
    from xgboost import XGBClassifier

    models["XGBoost"] = make_pipeline(
        XGBClassifier(
            n_estimators=250,
            max_depth=4,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            reg_lambda=5,
            tree_method="hist",
            eval_metric="logloss",
            n_jobs=4,
            random_state=42,
        )
    )
except ImportError:
    print(
        "\nXGBoost is not installed; other models will run."
        "\nTo add it later, run: %pip install xgboost"
    )

# =====================================================
# 5. FIT AND EVALUATE ALL MODELS
# =====================================================

fitted_models = {}
records = []

training_prevalence = y_train.mean()

for name, estimator in models.items():
    print(f"\nTraining {name}...")
    start = time.perf_counter()

    estimator.fit(X_train, y_train)
    fitted_models[name] = estimator

    elapsed = time.perf_counter() - start
    print(f"Fit time: {elapsed:.1f} seconds")

    for period_name, split in splits.items():
        if split.empty:
            continue

        X = split[features]
        y = split[target].astype(int)

        probabilities = estimator.predict_proba(X)[:, 1]

        top_n = max(1, int(np.ceil(0.10 * len(y))))
        top_idx = np.argsort(probabilities)[-top_n:]
        top_decile_rate = y.iloc[top_idx].mean()

        records.append({
            "model": name,
            "period": period_name,
            "rows": len(y),
            "delay_rate": y.mean(),
            "PR_AUC": average_precision_score(y, probabilities),
            "ROC_AUC": roc_auc_score(y, probabilities),
            "Brier": brier_score_loss(y, probabilities),
            "Log_Loss": log_loss(y, probabilities),
            "Top_10pct_Delay_Rate": top_decile_rate,
            "Top_10pct_Lift": top_decile_rate / y.mean(),
            "Training_Prevalence_Brier": brier_score_loss(
                y, np.full(len(y), training_prevalence)
            ),
            "fit_seconds": elapsed,
        })

results = pd.DataFrame(records)

# =====================================================
# 6. DISPLAY COMPARISONS
# =====================================================

metric_cols = [
    "PR_AUC",
    "ROC_AUC",
    "Brier",
    "Log_Loss",
    "Top_10pct_Delay_Rate",
    "Top_10pct_Lift",
]

for period_name in splits:
    print(f"\n{'=' * 20} {period_name} {'=' * 20}")

    table = (
        results.loc[results["period"].eq(period_name)]
        .sort_values("PR_AUC", ascending=False)
        [["model"] + metric_cols + ["Training_Prevalence_Brier"]]
    )

    print(table.round(4).to_string(index=False))

print("\nBenchmark complete.")
```

```python
import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
)

# ---------------------------------------------------
# 1. Prepare the dataset
# ---------------------------------------------------

bench = enhanced.copy()

bench["created_date"] = pd.to_datetime(
    bench["created_date"], errors="coerce"
)

bench["delay_gt_7d"] = pd.to_numeric(
    bench["delay_gt_7d"], errors="coerce"
)

bench = bench.dropna(
    subset=["created_date", "delay_gt_7d"]
).copy()

bench["created_month_cat"] = (
    bench["created_date"].dt.month.astype(str)
)
bench["created_weekday_cat"] = (
    bench["created_date"].dt.dayofweek.astype(str)
)
bench["created_hour"] = bench["created_date"].dt.hour
bench["year"] = bench["created_date"].dt.year

categorical_features = [
    "problem_formerly_complaint_type",
    "problem_detail_formerly_descriptor",
    "borough_clean",
    "incident_zip",
    "open_data_channel_type",
    "created_month_cat",
    "created_weekday_cat",
]

numeric_features = [
    "created_hour",
    "borough_reports_prev_7d",
    "borough_reports_prev_28d",
    "zip_reports_prev_7d",
    "zip_reports_prev_28d",
    "city_reports_prev_7d",
    "city_reports_prev_28d",
]

categorical_features = [
    c for c in categorical_features if c in bench.columns
]
numeric_features = [
    c for c in numeric_features if c in bench.columns
]

full_features = categorical_features + numeric_features

for col in categorical_features:
    bench[col] = (
        bench[col].astype("string").fillna("Unknown").astype(str)
    )

for col in numeric_features:
    bench[col] = pd.to_numeric(bench[col], errors="coerce")

# ---------------------------------------------------
# 2. Small, interpretable parameter search
# ---------------------------------------------------

configs = [
    {
        "name": "Current baseline",
        "max_depth": 20,
        "min_samples_leaf": 30,
        "max_features": "sqrt",
    },
    {
        "name": "Shallower trees",
        "max_depth": 12,
        "min_samples_leaf": 30,
        "max_features": "sqrt",
    },
    {
        "name": "Deeper trees",
        "max_depth": 24,
        "min_samples_leaf": 30,
        "max_features": "sqrt",
    },
    {
        "name": "Larger leaves",
        "max_depth": 20,
        "min_samples_leaf": 60,
        "max_features": "sqrt",
    },
    {
        "name": "Smaller leaves",
        "max_depth": 20,
        "min_samples_leaf": 10,
        "max_features": "sqrt",
    },
    {
        "name": "More features per split",
        "max_depth": 20,
        "min_samples_leaf": 30,
        "max_features": 0.7,
    },
]

# ---------------------------------------------------
# 3. Shared preprocessing
# ---------------------------------------------------

preprocessor = ColumnTransformer([
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]), categorical_features),

    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
    ]), numeric_features),
])

# ---------------------------------------------------
# 4. Expanding-window validation
# ---------------------------------------------------

results = []

for validation_year in [2022, 2023, 2024]:

    train = bench.loc[
        bench["year"].between(2020, validation_year - 1)
    ]
    valid = bench.loc[
        bench["year"].eq(validation_year)
    ]

    X_train = train[full_features]
    y_train = train["delay_gt_7d"].astype(int)

    X_valid = valid[full_features]
    y_valid = valid["delay_gt_7d"].astype(int)

    baseline_brier = brier_score_loss(
        y_valid,
        np.full(len(y_valid), y_train.mean())
    )

    print(
        f"Validating {validation_year}: "
        f"{len(train):,} train / {len(valid):,} validation"
    )

    for cfg in configs:
        model = Pipeline([
            ("preprocess", preprocessor),
            ("classifier", ExtraTreesClassifier(
                n_estimators=150,
                max_depth=cfg["max_depth"],
                min_samples_leaf=cfg["min_samples_leaf"],
                max_features=cfg["max_features"],
                n_jobs=-1,
                random_state=42,
            )),
        ])

        start = time.perf_counter()
        model.fit(X_train, y_train)
        fit_seconds = time.perf_counter() - start

        probabilities = model.predict_proba(X_valid)[:, 1]
        brier = brier_score_loss(y_valid, probabilities)

        results.append({
            "validation_year": validation_year,
            "configuration": cfg["name"],
            "PR_AUC": average_precision_score(
                y_valid, probabilities
            ),
            "ROC_AUC": roc_auc_score(
                y_valid, probabilities
            ),
            "Brier": brier,
            "Brier_skill": (
                1 - brier / baseline_brier
                if baseline_brier > 0 else np.nan
            ),
            "fit_seconds": fit_seconds,
        })

results_tuning = pd.DataFrame(results)

print("\nRESULTS BY YEAR")
print(
    results_tuning.sort_values(
        ["validation_year", "PR_AUC"],
        ascending=[True, False]
    ).round(4).to_string(index=False)
)

print("\nAVERAGE RESULTS")
print(
    results_tuning.groupby("configuration").agg(
        mean_PR_AUC=("PR_AUC", "mean"),
        mean_ROC_AUC=("ROC_AUC", "mean"),
        mean_Brier=("Brier", "mean"),
        mean_Brier_skill=("Brier_skill", "mean"),
        mean_fit_seconds=("fit_seconds", "mean"),
    ).sort_values("mean_PR_AUC", ascending=False)
    .round(4)
    .to_string()
)
```

```python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
)

# Prepare data
d = enhanced.copy()
d["created_date"] = pd.to_datetime(d["created_date"], errors="coerce")
d["year"] = d["created_date"].dt.year
d["delay_gt_7d"] = d["delay_gt_7d"].astype(int)

categorical = [
    "problem_formerly_complaint_type",
    "problem_detail_formerly_descriptor",
    "borough_clean",
    "incident_zip",
    "open_data_channel_type",
    "created_month_cat",
    "created_weekday_cat",
]

numeric = [
    "created_hour",
    "borough_reports_prev_7d",
    "borough_reports_prev_28d",
    "zip_reports_prev_7d",
    "zip_reports_prev_28d",
    "city_reports_prev_7d",
    "city_reports_prev_28d",
]

categorical = [c for c in categorical if c in d.columns]
numeric = [c for c in numeric if c in d.columns]
features = categorical + numeric

for c in categorical:
    d[c] = d[c].astype("string").fillna("Unknown").astype(str)

for c in numeric:
    d[c] = pd.to_numeric(d[c], errors="coerce")

# Fit on 2020–2024
train = d[d["year"].between(2020, 2024)].copy()

# Later periods are diagnostic only
test_2025 = d[d["year"] == 2025].copy()
test_2026 = d[d["year"] == 2026].copy()

preprocessor = ColumnTransformer([
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]), categorical),
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
    ]), numeric),
])

final_candidate = Pipeline([
    ("preprocess", preprocessor),
    ("model", ExtraTreesClassifier(
        n_estimators=150,
        max_depth=12,
        min_samples_leaf=30,
        max_features="sqrt",
        n_jobs=-1,
        random_state=42,
    )),
])

final_candidate.fit(
    train[features],
    train["delay_gt_7d"],
)

for name, split in [
    ("2025 diagnostic", test_2025),
    ("2026 stress diagnostic", test_2026),
]:
    X = split[features]
    y = split["delay_gt_7d"]
    p = final_candidate.predict_proba(X)[:, 1]

    baseline = np.full(len(y), train["delay_gt_7d"].mean())

    print(f"\n{name}")
    print("Rows:", len(split))
    print("Actual delay rate:", round(y.mean(), 4))
    print("PR-AUC:", round(average_precision_score(y, p), 4))
    print("ROC-AUC:", round(roc_auc_score(y, p), 4))
    print("Brier:", round(brier_score_loss(y, p), 4))
    print(
        "Constant-baseline Brier:",
        round(brier_score_loss(y, baseline), 4),
    )
```

```python
import joblib
from pathlib import Path
from datetime import date

artifact = {
    "model": final_candidate,
    "model_name": "Extra Trees",
    "model_version": "servicelens-v1",
    "feature_columns": features,
    "categorical_features": categorical,
    "numeric_features": numeric,
    "target": "delay_gt_7d",
    "target_definition": (
        "Administrative closure took more than 7 days "
        "or the request was still open after 7 days."
    ),
    "training_period": "2020-2024",
    "trained_on_date": date.today().isoformat(),
    "random_state": 42,
}

output_path = Path("servicelens_v1.joblib")
joblib.dump(artifact, output_path)

print(f"Saved model to: {output_path.resolve()}")
print(f"Features expected: {len(features)}")
print(f"Model size: {output_path.stat().st_size / 1024:.1f} KB")
```
